```python
class Actor(nn.Module):
    """
    Randomly initialize the Actor πθ
    Thus assuming standard implementation.
    
    maps: state -> action
    From paper: "By two ReLU layers and one Tanh layer, the state representation s 
    is transformed into an action a = π_θ(s)"
    """
    def __init__(self, state_dim, action_dim=100, hidden_dim=256):
        super().__init__()
        self.fc1 = nn.Linear(state_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, hidden_dim)
        self.fc3 = nn.Linear(hidden_dim, action_dim)
        
        # Initialize weights
        nn.init.xavier_uniform_(self.fc1.weight)
        nn.init.xavier_uniform_(self.fc2.weight)
        nn.init.xavier_uniform_(self.fc3.weight)
        
    def forward(self, state):
        x = F.relu(self.fc1(state))
        x = F.relu(self.fc2(x))
        # Tanh maps action to [-1, 1]
        action = torch.tanh(self.fc3(x))
        return action
```

```python
# Load Actor model
version_number = 2

state_dim = 300
actor = Actor(state_dim)
actor = actor.to(device)
# Load the weights into that architecture
actor.load_state_dict(torch.load(f'drr_actor_weights_{version_number}.pth'))
actor.eval()
```

```python
# Actor takes in state, outputs an action.
# Each action is a continuous 1xk row vector representing a "movie" in latent space
# The top ranked movies are recommended to the user (according to epsilon greedy)
# item_embeddings: all movies in MovieLens 1M
# epsilon set to 0.1 is standard practice
# num items: all unique movies in MovieLens 1M
# 3533 movies because
# 1. I only kept userIDs who rated movies 4 or 5
# 2. I then only kept ALL movies THOSE users rated
# embed_dim = k = 100
class ItemRanker:
    """
    Ranks items based on action vector and item embeddings
    Implements Eq. 2: score = item_embedding · action^T
    """
    def __init__(self, item_embeddings, epsilon):
        """
        Args:
            item_embeddings: numpy array of shape (num_items, embed_dim)
            epsilon: exploration rate for ε-greedy
        """
        self.item_embeddings = torch.FloatTensor(item_embeddings)
        self.epsilon = epsilon
        self.num_items = item_embeddings.shape[0]
        self.embed_dim = item_embeddings.shape[1]
        
    def get_all_scores(self, action):
        """
        Compute scores for all items
        Args:
            action: (batch_size, embed_dim) action vector
        Returns:
            scores: (batch_size, num_items) scores for all items
        """
        # Dot product: (batch_size, embed_dim) @ (embed_dim, num_items)
        scores = action @ self.item_embeddings.T  # (batch_size, num_items)
        return scores
    
    def select_item(self, action, candidate_items=None, exclude_items=None):
        """
        Select an item using ε-greedy
        Args:
            action: (embed_dim,) action vector for a single user
            candidate_items: list of candidate item indices (if None, use all items)
            exclude_items: set of items to exclude (already recommended)
        Returns:
            selected_item: index of selected item
        """
        action = action.squeeze()  # Ensure 1D
        
        # Get candidate items
        if candidate_items is None:
            candidate_items = list(range(self.num_items))
        
        # Filter out excluded items
        if exclude_items:
            filtered_items = []
            for i in candidate_items:
                if i not in exclude_items:
                    filtered_items.append(i)            
            candidate_items = filtered_items
        
        if not candidate_items:
            return None
        
        # ε-greedy exploration
        if np.random.random() < self.epsilon:
            # If it meets the threshold, Explore: random item
            return np.random.choice(candidate_items)
        else:
            candidate_embeddings = self.item_embeddings[candidate_items].cpu().numpy()
            scores = action @ candidate_embeddings.T
            best_idx = np.argmax(scores)
            return candidate_items[best_idx]
```

```python
def explore_test_set(test_sequences, user2idx, n_history):
    missing_users = []
    users_with_not_enough_history = []
    empty_user_test_sets = []
    users_with_no_positive_ground_truths = []

    
    for user_id in test_sequences.keys():
        # Get user ID index
        if user_id not in user2idx: 
            missing_users.append(user_id)
            continue
        user_idx = user2idx[user_id]
        
        # Build INITIAL STATE from TRAIN
        train_seq = train_sequences.get(user_id, [])
        train_positive = []
        for (m, r) in train_seq:
            if r >= 4 and m in movie2idx:
                # Convert movie ID to its index representation
                movie_index = movie2idx[m]
                train_positive.append(movie_index)
        if len(train_positive) < n_history: 
            # print(f"{user_id} has a short past of {len(train_positive)}")
            users_with_not_enough_history.append(user_id)
            continue
        history = train_positive[-n_history:].copy()

        # Candidate set = TEST items
        test_seq = test_sequences[user_id]
        candidate_items = []
        for (m, _) in test_seq:
            if m in movie2idx:
                # Convert movie ID to index
                movie_index = movie2idx[m]
                candidate_items.append(movie_index)
        if len(candidate_items) == 0: 
            empty_user_test_sets.append(user_id)
            continue


        # Ground truth for metrics (positive test items)
        ground_truth = set()
        for (m, r) in test_seq:
            if r >= 4 and m in movie2idx:
                # Convert movie ID to index
                movie_index = movie2idx[m]
                ground_truth.add(movie_index)
        if len(ground_truth) == 0: 
            users_with_no_positive_ground_truths.append(user_id)
            continue

    print(f"# of missing users: {len(missing_users)}")
    print(f"# of users with not enough train set past: {len(users_with_not_enough_history)}")
    print(f"# of users with NO rated movies: {len(empty_user_test_sets)}")
    print(f"# of users with no positive ground truths: {len(users_with_no_positive_ground_truths)}")
```

# Algorithm 2

```python
def get_state(user_idx, history_indices, user_embeddings, movie_embeddings, item_weights=None):
    """
    DRR-ave state for evaluation: s = [u, u⊗g, g]
    
    Args:
        user_idx: index of user
        history_indices: list of item indices (length n_history)
        user_embeddings: array of shape (num_users, embed_dim)
        movie_embeddings: array of shape (num_movies, embed_dim)
        item_weights: optional learned weights (if None, use uniform weights)
    
    Returns:
        state: concatenated array of shape (3*embed_dim,)
    """
    # Get user embedding
    u = user_embeddings[user_idx]  # (embed_dim,)
    
    # Get item embeddings for history
    item_embs = movie_embeddings[history_indices]  # (n_history, embed_dim)
    
    # Apply weights if provided (for evaluation with trained weights)
    if item_weights is not None:
        # Softmax to match training
        weights = np.exp(item_weights) / np.sum(np.exp(item_weights))
        weighted_embs = item_embs * weights[:, np.newaxis]
        g = np.sum(weighted_embs, axis=0)
    else:
        # Simple average (fallback)
        g = np.mean(item_embs, axis=0)
    
    # Element-wise product
    u_g_interaction = u * g
    
    # Concatenate
    state = np.concatenate([u, u_g_interaction, g])
    
    return state.astype(np.float32)
```

```python
def get_user_history(train_sequences, user_id, movie2idx, n_history=5, min_rating=4):
    """
    Extract and prepare user's training history.
    
    Returns:
        list: User's recent positive movie indices, or None if insufficient history
    """    
    # train_seq is a List of tuples (movieID, rating) for userID ONLY
    train_seq = train_sequences.get(user_id, [])
    train_positive = []
    
    for movieID, rating in train_seq:
        if rating >= min_rating and movieID in movie2idx:
            # Convert movie ID to its index representation
            movie_index = movie2idx[movieID]
            train_positive.append(movie_index)
    
    if len(train_positive) < n_history:
        return None

    history = train_positive[-n_history:].copy()
    return history
```

```python
def get_session_ratings(test_seq, movie2idx):
    """
    Create a dictionary mapping movie indices to ratings from test sequence.
    
    Args:
        test_seq: List of (movie_id, rating) tuples for a user
        movie2idx: Dictionary mapping movie IDs to indices
    
    Returns:
        dict: Dictionary with movie indices as keys and ratings as values
    """
    session_ratings = {}
    for movie_id, rating in test_seq:
        if movie_id in movie2idx:
            session_ratings[movie2idx[movie_id]] = rating
    return session_ratings
```

```python
def get_ground_truth(test_seq, movie2idx, rating_threshold=4):
    """
    Create set of movie indices that user rated highly (ground truth for recommendations).
    
    Args:
        test_seq: List of (movie_id, rating) tuples for a user
        movie2idx: Dictionary mapping movie IDs to indices
        rating_threshold: Minimum rating to consider as positive (default: 4)
    
    Returns:
        set: Set of movie indices with ratings >= threshold, or None if empty
    """
    ground_truth = set()
    for movie_id, rating in test_seq:
        if rating >= rating_threshold and movie_id in movie2idx:
            movie_index = movie2idx[movie_id]
            ground_truth.add(movie_index)
    
    if len(ground_truth) == 0:
        return None
    
    return ground_truth
```

```python
def algo_2(user_idx, history, candidate_items, session_ratings, 
               user_embeddings, movie_embeddings, actor, item_ranker, 
               device,
               n_history=5, max_episode_length=10):
    """
    Evaluation algorithm

    Returns list of recommended items, and list of scores for each item
    """
    recommended_items = []
    recommended_scores = []
    
    for t in range(min(max_episode_length, len(candidate_items))):
        # Get current state
        state = get_state(user_idx, history[-n_history:], user_embeddings, movie_embeddings)
        state_tensor = torch.FloatTensor(state).unsqueeze(0).to(device)

        # Execute action according to optimal policy
        with torch.no_grad():
            action = actor(state_tensor)
        action_np = action.squeeze().cpu().numpy()

        # Select best item (highest score; score = item_embedding · action)
        selected_item = item_ranker.select_item(action_np, candidate_items, set(recommended_items))
        if selected_item is None: break

        # Calculate the "Predicted Score" manually for display
        item_emb = movie_embeddings[selected_item]
        score = np.dot(action_np, item_emb)
        
        recommended_items.append(selected_item)
        recommended_scores.append(score)
        candidate_items.remove(selected_item)

        # Reward from TEST log (Eq 10)
        rating = None
        rating = session_ratings[selected_item]
        if rating is None: continue
        
        reward = 0.5 * (rating - 3)
        # Update state ONLY if reward > 0s
        if reward > 0:
            history.append(selected_item)

    return recommended_items, recommended_scores
```

```python
def print_user_recommendations(user_id, user_idx, test_seq, ground_truth, 
                               recommended_items, recommended_scores, 
                               idx2movie, movie_id_to_title, rating_threshold=4):
    """
    Print detailed recommendation results for a single user.
    
    Args:
        user_id: Original user ID
        user_idx: Index of the user in the embedding matrix
        test_seq: List of (movie_id, rating) tuples for the user's test sequence
        ground_truth: Set of movie indices that are considered ground truth (ratings >= threshold)
        recommended_items: List of recommended movie indices
        recommended_scores: List of scores corresponding to recommended_items
        idx2movie: Dictionary mapping movie indices to original movie IDs
        movie_id_to_title: Dictionary mapping movie IDs to titles
        rating_threshold: Minimum rating to consider as favorite (default: 4)
    """
    print(f"\n" + "="*70)
    print(f"USER ID: {user_id} (Index: {user_idx})")
    
    # Create rating lookup from test_seq
    rating_lookup = {}
    for movie_id, rating in test_seq:
        if movie_id in idx2movie.values(): 
            rating_lookup[movie_id] = rating
            
    # Print ground truth favorites from test set
    print(f"Test Set Favorites (Ground Truth):")
    for movie_id, rating in test_seq:
        if rating >= rating_threshold:
            title = movie_id_to_title.get(movie_id, 'Unknown')
            print(f"  - {title} (Actual Rating: {rating})")

    recommendation_pairs = list(zip(recommended_items, recommended_scores))
    
    # Print recommendations table
    print(f"\nModel Recommendations:")
    print(f"{'Rank':<5} | {'Title':<40} | {'Score':<10} | {'Hit?':<6} | {'Actual Rating':<6}")
    print("-" * 80)
    
    for i, (movie_idx, score) in enumerate(recommendation_pairs):
        movie_id = idx2movie[movie_idx]
        title = movie_id_to_title.get(movie_id, "Unknown")
        is_hit = "YES" if movie_idx in ground_truth else "NO"
        actual_rating = rating_lookup.get(movie_id, "N/A")
        print(f"{i+1:<5} | {title[:40]:<40} | {score: <10.4f}  | {is_hit:<6} | {actual_rating:<6}")
    
    print("="*80)
```

```python
def compute_metrics_at_k(recommended_items, ground_truth, k):
    """
    Compute Precision@k and NDCG@k for a single user.
    
    Args:
        recommended_items: List of recommended movie indices in order
        ground_truth: Set of movie indices that are relevant (highly rated)
        k: Number of top recommendations to consider
    
    Returns:
        tuple: (precision_at_k, ndcg_at_k)
    """
    if len(recommended_items) == 0:
        return 0, 0
    
    recommended_at_k = recommended_items[:k]
    
    # Precision@K
    hits = len(set(recommended_at_k) & ground_truth)
    precision_at_k = hits / float(k)
    
    # NDCG@K
    dcg = 0
    for i, item in enumerate(recommended_at_k):
        if item in ground_truth:
            dcg += 1 / np.log2(i + 2)
    
    idcg = 0.0
    max_rank = min(len(ground_truth), k)
    for i in range(max_rank):
        discounted_gain = 1 / np.log2(i + 2)                    
        idcg += discounted_gain

    ndcg_at_k = 0
    if idcg > 0:
        ndcg_at_k = dcg / idcg 
    
    return precision_at_k, ndcg_at_k
```

```python
def get_results(precision, ndcg, top_k_list):
    """
    Aggregate precision and NDCG scores across all users into final results.
    
    Args:
        precision: Dictionary mapping k to list of precision scores for each user
        ndcg: Dictionary mapping k to list of NDCG scores for each user
        top_k_list: List of k values to compute metrics for
    
    Returns:
        dict: Dictionary with metric names as keys and mean scores as values
    """
    results = {}
    for k in top_k_list:
        # Precision@K
        precision_mean = 0
        # Get the list of precision scores for this k
        precision_list = precision[k]
        if precision_list:  
            # if list is not empty
            precision_mean = np.mean(precision_list)
        results[f'Precision@{k}'] = precision_mean
        
        # NDCG@K
        ndcg_mean = 0
        # Get the list of NDCG scores for this k
        ndcg_list = ndcg[k]
        if ndcg_list:  
            # if list is not empty
            ndcg_mean = np.mean(ndcg_list)
            
        results[f'NDCG@{k}'] = ndcg_mean

    return results
```

# Eval model

```python
def evaluate_drr_with_trajectory(train_sequences, test_sequences, 
                                 user2idx, movie2idx, idx2movie,
                                 user_embeddings, movie_embeddings, 
                                 actor, item_ranker, max_episode_length=10,
                                 n_history=5, top_k_list=[5, 10]):
    
    # Store all session trajectories
    all_sessions = {}
    
    # Loop through users
    all_user_ids = list(test_sequences.keys())
    for user_id in all_user_ids:
        if user_id not in user2idx: continue
        user_idx = user2idx[user_id]
        
        # Get initial history from training
        history = get_user_history(train_sequences, user_id, movie2idx, n_history)
        if history is None: continue
        
        test_seq = test_sequences[user_id]
        session_ratings = get_session_ratings(test_seq, movie2idx)
        candidate_items = list(session_ratings.keys())
        
        ground_truth = get_ground_truth(test_seq, movie2idx, rating_threshold=4)
        if ground_truth is None: continue
        
        # Run the session and SAVE the trajectory
        trajectory = {
            'user_id': user_id,
            'user_idx': user_idx,
            'initial_history': history.copy(),
            'steps': [],
            'ground_truth': ground_truth,
            'test_seq': test_seq
        }
        
        recommended_items = []
        recommended_scores = []
        current_history = history.copy()
        
        for t in range(min(max_episode_length, len(candidate_items))):
            # Get state
            state = get_state(user_idx, current_history[-n_history:], user_embeddings, movie_embeddings)
            state_tensor = torch.FloatTensor(state).unsqueeze(0).to(device)
            
            with torch.no_grad():
                action = actor(state_tensor)
            action_np = action.squeeze().cpu().numpy()
            
            # Select item
            selected_item = item_ranker.select_item(action_np, candidate_items, set(recommended_items))
            if selected_item is None: break
            
            # Calculate score
            item_emb = movie_embeddings[selected_item]
            score = np.dot(action_np, item_emb)
            
            recommended_items.append(selected_item)
            recommended_scores.append(score)
            candidate_items.remove(selected_item)
            
            # Get reward
            rating = session_ratings[selected_item]
            reward = 0.5 * (rating - 3)
            
            # SAVE THIS STEP
            step_data = {
                'step': t+1,
                'recommended_movie_idx': selected_item,
                'recommended_movie_title': movie_id_to_title[idx2movie[selected_item]],
                'rating': rating,
                'score': score,
                'reward': reward,
                'history_at_time': current_history.copy()  # History used for this recommendation
            }
            trajectory['steps'].append(step_data)
            
            # Update history if reward > 0
            if reward > 0:
                current_history.append(selected_item)
        
        # Save all user sessions to 1 dictionary key=userID, value = trajectory
        # trajectory is a dictionary with lots of key-value pairs to later do SHAP explanations
        all_sessions[user_id] = trajectory
        
        # Print as before
        print_user_recommendations(user_id, user_idx, test_seq, ground_truth, 
                                   recommended_items, recommended_scores, 
                                   idx2movie, movie_id_to_title, rating_threshold=4)

    # Calculate metrics
    precision = {}
    ndcg = {}
    for k in top_k_list:
        precision[k] = []
        ndcg[k] = []
    
    for user_id, session in all_sessions.items():
        # Get recommended items from the trajectory
        recommended_items = [step['recommended_movie_idx'] for step in session['steps']]
        ground_truth = session['ground_truth']
        
        if len(recommended_items) > 0:
            for k in top_k_list:
                precision_at_k, ndcg_at_k = compute_metrics_at_k(recommended_items, ground_truth, k)
                precision[k].append(precision_at_k)
                ndcg[k].append(ndcg_at_k)
    
    results = get_results(precision, ndcg, top_k_list)
    print("\nEvaluation Results:")
    for metric, value in results.items():
        print(f"{metric}: {value:.4f}")
    
    return all_sessions, results
```

```python
item_ranker = ItemRanker(movie_embeddings, epsilon=0.0)
```

```python
sessions, results = evaluate_drr_with_trajectory(
    train_sequences, test_sequences, 
    user2idx, movie2idx, idx2movie,
    user_embeddings, movie_embeddings, 
    actor, item_ranker, max_episode_length=10,
    n_history=5, top_k_list=[5, 10]
)

# Save the trajectory
with open('session_trajectories.pkl', 'wb') as f:
    pickle.dump(sessions, f)
```

```python
print("Evaluation on Test Set:")
for metric, value in results.items():
    print(f"{metric}: {value:.4f}")

# Evaluation on Test Set:
# Precision@5: 0.6025
# NDCG@5: 0.6320
# Precision@10: 0.5448
# NDCG@10: 0.6753
```
